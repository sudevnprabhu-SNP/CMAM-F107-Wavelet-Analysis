# 01 — Project Data Overview

This notebook checks the input files and inspects the schemas before analysis. It does not modify the raw data.

**Expected inputs**
- `data/raw/cmam/2005.nc` through `2010.nc`
- `data/raw/f107/F107_daily_1979_2010.csv`

Run from the repository root or adjust `PROJECT_ROOT` below.


```python
from pathlib import Path
import pandas as pd
import xarray as xr

PROJECT_ROOT = Path.cwd()
# If opened from notebooks/, use this fallback:
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

CMAM_DIR = PROJECT_ROOT / "data" / "raw" / "cmam"
F107_FILE = PROJECT_ROOT / "data" / "raw" / "f107" / "F107_daily_1979_2010.csv"

print("Project root:", PROJECT_ROOT)
print("CMAM directory exists:", CMAM_DIR.exists())
print("F10.7 file exists:", F107_FILE.exists())
```

```python
# Inspect each CMAM file's dimensions, variables, calendar, and pressure units.
for year in range(2005, 2011):
    path = CMAM_DIR / f"{year}.nc"
    if not path.exists():
        print("Missing:", path)
        continue
    with xr.open_dataset(path) as ds:
        print(f"\\n{year}: dimensions={dict(ds.sizes)}")
        print("Variables:", list(ds.data_vars))
        print("plev units:", ds["plev"].attrs.get("units", "not specified"))
        print("ta units:", ds["ta"].attrs.get("units", "not specified"))
        print("Calendar:", ds["time"].encoding.get("calendar", "default/unspecified"))
```

```python
# Inspect the F10.7 CSV without assuming that it has been cleaned.
if F107_FILE.exists():
    solar = pd.read_csv(F107_FILE)
    print("Columns:", solar.columns.tolist())
    display(solar.head())
    print("Rows:", len(solar))
else:
    print("Place F107_daily_1979_2010.csv in data/raw/f107/ before continuing.")
```
