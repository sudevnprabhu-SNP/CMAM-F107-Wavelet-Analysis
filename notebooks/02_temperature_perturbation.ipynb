# 02 — CMAM Temperature Perturbation

This notebook inspects the processed monthly temperature series at approximately 120 km and 0° latitude. It expects the preparation script to have created `data/processed/temperature_perturbation.csv`.

The reference mean was calculated over the available monthly interval; perturbations are temperature minus that reference mean.


```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "data").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent
CSV = PROJECT_ROOT / "data" / "processed" / "temperature_perturbation.csv"

if not CSV.exists():
    raise FileNotFoundError(f"{CSV} not found. Run src/temperature/02_prepare_monthly_series.py first.")

df = pd.read_csv(CSV)
df.head()
```

```python
df["date"] = pd.to_datetime(df["date"])
value_col = next(c for c in ["temperature_perturbation_K", "temperature_perturbation"] if c in df.columns)
df[value_col] = pd.to_numeric(df[value_col], errors="raise")
df = df.sort_values("date")

print("Rows:", len(df))
print("Date range:", df["date"].min(), "to", df["date"].max())
print("Mean perturbation (K):", df[value_col].mean())
print("Minimum / maximum (K):", df[value_col].min(), df[value_col].max())
```

```python
fig, ax = plt.subplots(figsize=(13, 4))
ax.plot(df["date"], df[value_col], marker="o", linewidth=1)
ax.axhline(0, color="black", linewidth=1)
ax.set_xlabel("Date")
ax.set_ylabel("Temperature perturbation (K)")
ax.set_title("Monthly CMAM temperature perturbation near 120 km, 0° latitude")
ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

## Interpretation

The zero line represents the reference-period mean. Positive values indicate temperatures above that reference mean; negative values indicate temperatures below it. The time average should be close to zero by construction. This series is only 66 months long, which limits reliable interpretation of multi-year variability.
